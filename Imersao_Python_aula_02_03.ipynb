Aula 2 - Importação

importando biblioteca pandas

```python
import pandas as pd
import plotly.express as px
```

carregando tabela principal e guardando em variavel dataframe

```python
df_principal = pd.read_excel("/content/acoes_puras.xlsx", sheet_name="Principal")
df_principal
```

```python
df_Total_de_acoes = pd.read_excel("/content/acoes_puras.xlsx", sheet_name="Total_de_acoes")
df_Total_de_acoes
```

```python
df_Ticker = pd.read_excel("/content/acoes_puras.xlsx", sheet_name="Ticker")
df_Ticker
```

```python
df_chatgpt = pd.read_excel("/content/acoes_puras.xlsx", sheet_name="chatgpt")
df_chatgpt
```

utilizando apenas as primeiras 4 colunas no df_principal

```python
df_principal = df_principal[['Ativo', 'Data', 'Último (R$)', 'Var. Dia (%)']].copy()
df_principal
```

renomeando as colunas no df_principal

```python
df_principal = df_principal.rename(columns={'Último (R$)':'valor_final','Var. Dia (%)': 'var_dia_pct'}).copy()
df_principal
```

adicionando coluna calculando valor _inicial
inicial = final/ 1 + var_dia_pct

```python
df_principal['var_dia_pct'] = df_principal['var_dia_pct']/100
df_principal['valor_inicial'] = df_principal['valor_final'] / (df_principal['var_dia_pct'] + 1)
df_principal
```

fazendo merge df_principal com df_Total_de_acoes
em comum ativo e código, informação importante qrt teorica

```python
df_principal = df_principal.merge(df_Total_de_acoes,left_on='Ativo',right_on='Código',how='left')
df_principal
```

retirando coluna codigo - pois esta duplicada

```python
df_principal = df_principal.drop(columns='Código').copy()
df_principal
```

adicional coluna variacao_rs =  final - inicial vezes qtde teorica

```python
df_principal['variacao_rs'] = (df_principal['valor_final']-df_principal['valor_inicial'])*df_principal['Qtde. Teórica']
df_principal
```

formatando valores como float - ponto flutuantes (com casas decimais)

```python
pd.options.display.float_format = '{:.2f}'.format
```

formatando Qtde. Teórica como numero inteiro

```python
df_principal['Qtde. Teórica'] = df_principal['Qtde. Teórica'].astype(int)
df_principal
```

renomeando coluna Qtde. Teórica para qtd_teorica

```python
df_principal = df_principal.rename(columns={'Qtde. Teórica':'Qtd_teorica'}).copy()
df_principal
```

adicionando coluna resultado
apply usa outras funcoes
lambda - funcao generica
em cada linha é aplicado a condicão if para classificação

```python
df_principal['Resultado'] = df_principal['variacao_rs'].apply(lambda x: 'Subiu' if x > 0 else ('Desceu' if x < 0 else 'Estável'))
df_principal
```

```python
df_principal = df_principal.merge(df_Ticker, left_on='Ativo', right_on='Ticker', how='left')
df_principal = df_principal.drop(columns=['Ticker'])
df_principal
```

```python
df_principal = df_principal.merge(df_chatgpt, left_on='Nome', right_on='Nome da Empresa', how='left')
df_principal = df_principal.drop(columns='Nome da Empresa').copy()
df_principal
```

```python
df_principal = df_principal.rename(columns={'Idade (anos)':'Ano_de_fundacao'}).copy()
df_principal
```

```python
df_principal['Cat_idade'] = df_principal['Ano_de_fundacao'].apply(lambda x: 'Mais de 100' if x > 100 else ('Menos de 50' if x < 50 else 'Entre 50 e 100'))
df_principal
```

Analises

```python
# Calculando o maior valor
maior = df_principal['variacao_rs'].max()

# Calculando o menor valor
menor = df_principal['variacao_rs'].min()

# Calculando a média
media = df_principal['variacao_rs'].mean()

# Calculando a média de quem subiu
media_subiu = df_principal[df_principal['Resultado'] == 'Subiu']['variacao_rs'].mean()

# Calculando a média de quem desceu
media_desceu = df_principal[df_principal['Resultado'] == 'Desceu']['variacao_rs'].mean()

# Imprimindo os resultados
print(f"Maior\tR$ {maior:,.2f}")
print(f"Menor\tR$ {menor:,.2f}")
print(f"Média\tR$ {media:,.2f}")
print(f"Média de quem subiu\tR$ {media_subiu:,.2f}")
print(f"Média de quem desceu\tR$ {media_desceu:,.2f}")
```

```python
df_principal_subiu = df_principal[df_principal['Resultado'] == 'Subiu']
df_principal_subiu
```

```python
df_analise_segmento = df_principal_subiu.groupby('Segmento')['variacao_rs'].sum().reset_index()
df_analise_segmento
```

```python
df_analise_saldo = df_principal.groupby('Resultado')['variacao_rs'].sum().reset_index()
df_analise_saldo
```

Criando grafico

```python
fig = px.bar(df_analise_saldo, x='Resultado', y='variacao_rs', text='variacao_rs', title='Variação Reais por Resultado')
fig.show()
```

```python
fig.update_traces(texttemplate="%{y:.2s}", textposition="outside")
fig.show()
```

```python
# Função para formatar os valores corretamente, incluindo números negativos
def formatar_valor(valor):
    abs_valor = abs(valor)  # Obtém o valor absoluto para comparação

    if abs_valor >= 1_000_000_000:
        return f"{valor / 1_000_000_000:.2f}B"  # Bilhões
    elif abs_valor >= 1_000_000:
        return f"{valor / 1_000_000:.2f}M"  # Milhões
    elif abs_valor >= 1_000:
        return f"{valor / 1_000:.2f}K"  # Milhares
    else:
        return f"{valor:.2f}"  # Valor normal

# Criando uma nova coluna com os valores formatados para rótulos
# Using df_analise_saldo instead of df and 'variacao_rs' instead of 'Valor'
df_analise_saldo["Valor_Formatado"] = df_analise_saldo["variacao_rs"].apply(formatar_valor)

# Criando o gráfico
# Using df_analise_saldo instead of df and appropriate column names
fig = px.bar(df_analise_saldo, x="Resultado", y="variacao_rs", text=df_analise_saldo["Valor_Formatado"])  # Usa a coluna formatada para os rótulos

# Posiciona os rótulos acima das barras
fig.update_traces(textposition="outside")

fig.show()
```

```python
fig = px.pie(df_analise_segmento, names='Segmento', values='variacao_rs', title='Variação Reais por Segmento')
fig.show()
```
